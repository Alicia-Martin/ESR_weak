```python
import pandas as pd
import numpy as np
import os

# -------- CONFIGURE PATHS --------
csv_path = 'bestfunc_comp10_params_per_cluster.csv'
# optional cluster names file (only used if you want to map ids to names)
names_path = '../all_clusters.txt'

# -------- SAFETY CHECKS --------
if not os.path.exists(csv_path):
    raise FileNotFoundError(f"CSV not found: {csv_path}")

# load CSV (skip comment lines beginning with '#')
df = pd.read_csv(csv_path, sep=';', comment='#', header=0, skipinitialspace=True)

print("Columns in CSV:", list(df.columns))

# -------- IDENTIFY COLUMNS --------
# find cluster id column (common names), fallback to first column
possible_id_names = ['cluster_id','id','name','cluster','cluster_name']
id_col = next((c for c in possible_id_names if c in df.columns), None)
if id_col is None:
    id_col = df.columns[0]
print("Using id column:", id_col)

# find likelihood column heuristically
lik_candidates = [c for c in df.columns if any(k in c.lower() for k in ('neglog','neg_log','negloglike','neglog_like','likelihood','total'))]
lik_col = lik_candidates[0] if lik_candidates else None
print("Using likelihood column:", lik_col)

# locate parameter columns that correspond to a0..a3
# Accept names like param_0, param0, a0, a_0 (case-insensitive)
def find_param_col(df_cols, target_idx):
    patterns = [f'param_{target_idx}', f'param{target_idx}', f'a{target_idx}', f'a_{target_idx}', f'param {target_idx}']
    lower = {c.lower(): c for c in df_cols}
    for p in patterns:
        if p in lower:
            return lower[p]
    # fallback: find nth 'param' column occurrence (param_0, param_1, ...)
    param_cols = [c for c in df_cols if c.lower().startswith('param')]
    if len(param_cols) > target_idx:
        return param_cols[target_idx]
    return None

param_cols_found = {}
for idx in range(0, 4):  # a0..a3
    colname = find_param_col(df.columns, idx)
    param_cols_found[idx] = colname
    print(f"a{idx} -> {colname}")

# Decide which column is a1 (the parameter used for filtering)
a1_col = param_cols_found.get(1)
if a1_col is None:
    raise ValueError("Could not identify a1/param_1 column. Found param columns: " + str(param_cols_found))

# coerce numeric
df[a1_col] = pd.to_numeric(df[a1_col], errors='coerce')
if lik_col:
    df[lik_col] = pd.to_numeric(df[lik_col], errors='coerce')

# -------- FILTER ROWS WHERE a1 IN (0.5, 5) --------
mask = (df[a1_col] > 0.5) & (df[a1_col] < 5)
filtered = df.loc[mask].copy()
print(f"Found {len(filtered)} rows with {a1_col} in (0.5, 5).")

# -------- PREPARE OUTPUT COLUMNS --------
out_cols = [id_col]
for idx in range(0, 4):
    c = param_cols_found.get(idx)
    out_cols.append(c if c is not None else f'a{idx}_missing')

if lik_col:
    out_cols.append(lik_col)

# Ensure missing columns are represented
for idx in range(0, 4):
    c = param_cols_found.get(idx)
    if c is None:
        filtered[f'a{idx}_missing'] = np.nan

# Reorder columns for printing/saving
out_cols = [id_col] + [ (param_cols_found[idx] if param_cols_found[idx] is not None else f'a{idx}_missing') for idx in range(4) ]
if lik_col:
    out_cols.append(lik_col)

# -------- PRINT RESULTS --------
print("\ncluster_id\t a0 \t a1 \t a2 \t a3 \t likelihood")
for _, row in filtered.iterrows():
    vals = [row[id_col]]
    for idx in range(4):
        col = param_cols_found.get(idx)
        v = row[col] if col is not None else np.nan
        vals.append(v)
    if lik_col:
        vals.append(row[lik_col])
    # pretty print
    print("\t".join(str(x) for x in vals))

# -------- SAVE TO CSV --------
out_dir = os.path.dirname(csv_path)
outpath = os.path.join(out_dir, 'filtered_params_a0_a3_a1range_0.5_5.csv')
filtered[out_cols].to_csv(outpath, index=False)
print("\nSaved filtered table to:", outpath)

# optional: print head of saved table
print("\nFirst few rows of saved table:")
print(filtered[out_cols].head().to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import os

# === Edit this path if your CSV lives elsewhere ===
csv_path = 'bestfunc_comp10_params_per_cluster.csv'

if not os.path.exists(csv_path):
    raise FileNotFoundError(f"CSV not found: {csv_path}")

# Read CSV (skip lines that start with '#')
df = pd.read_csv(csv_path, sep=';', comment='#', header=0, skipinitialspace=True)
# print(df)

print("Columns detected:", df.columns.tolist())

# Heuristically detect parameter columns (param_*, param*, a0,a1,...)
param_cols = []
# prefer explicit param_ or aX columns
for c in df.columns:
    cl = c.lower().replace(' ', '').replace('-', '_')
    if cl.startswith('param') or (cl.startswith('a') and len(cl) > 1 and cl[1].isdigit()):
        param_cols.append(c)

# fallback: any column with 'param' in the name
if not param_cols:
    param_cols = [c for c in df.columns if 'param' in c.lower()]

if not param_cols:
    raise ValueError("No parameter columns found. Columns are: " + ", ".join(df.columns))

print("Parameter columns inferred:", param_cols)

# Identify a0 column (common names)
a0_candidates = [c for c in df.columns if c.lower() in ('a0','a_0','param_0','param0')]
a0_col = a0_candidates[0] if a0_candidates else param_cols[0]
print("Using a0 column:", a0_col)

# identify id column (fall back to first column)
possible_id_names = ['cluster_id','id','name','cluster','cluster_name']
id_col = next((c for c in df.columns if c in possible_id_names), None)
if id_col is None:
    id_col = df.columns[0]
print("Using id column:", id_col)

# detect likely likelihood/neglog column
lik_cols = [c for c in df.columns if any(k in c.lower() for k in ('neglog','neg_log','likelihood','total','lik'))]
print("Possible likelihood columns:", lik_cols)

# convert parameter columns to numeric (coerce errors to NaN)
for c in param_cols:
    df[c] = pd.to_numeric(df[c], errors='coerce')

# convert likelihood if present
for c in lik_cols:
    df[c] = pd.to_numeric(df[c], errors='coerce')

# Filter: a0 in (-5, -0.5)
low, high = -5.0, -0.5
mask = (df[a0_col] > low) & (df[a0_col] < high)
filtered = df.loc[mask].copy()

print(f"\nNumber of rows with {a0_col} in ({low}, {high}): {len(filtered)}\n")

if len(filtered) == 0:
    print("No rows matched the filter. Check column detection or bounds.")
else:
    # Columns to show: id, all params, likelihoods (if any)
    cols_to_show = [id_col] + param_cols + lik_cols
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 140)
    print(filtered[cols_to_show].to_string(index=False))

    # Save filtered rows to CSV
    outdir = os.path.dirname(csv_path)
    outname = os.path.splitext(os.path.basename(csv_path))[0] + f'_filtered_a0_{int(abs(low))}_to_{int(abs(high))}.csv'
    outpath = os.path.join(outdir, outname)
    filtered[cols_to_show].to_csv(outpath, index=False)
    print("\nSaved filtered table to:", outpath)

```
